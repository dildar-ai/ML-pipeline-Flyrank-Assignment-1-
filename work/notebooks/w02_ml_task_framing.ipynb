# ML-03 — Frame Your Lane as an ML Task

## 1. My lane as an ML task (type)

I am framing this as a **binary classification** task. The goal is to predict which customers are likely to cancel their subscription (churn) versus those who will stay. Classification is appropriate because we are assigning a categorical label ('Churn' or 'No Churn') to each user based on their historical behavior and demographic features.

```python
# Example of class distribution check
import pandas as pd
import numpy as np

# Mock data to simulate checking class balance
target_counts = pd.Series([850, 150], index=['Stayed', 'Churned'])
print("Class Distribution:")
print(target_counts / target_counts.sum())
# A 15% churn rate suggests a need for handling class imbalance during training.
```

## 2. Target or proxy

**Target:** A binary indicator `is_churned`.
**Label Source:** This is an **observed outcome**. We define a 'churn' as a customer who has not had an active subscription for more than 30 days. We look back at historical billing logs to label users who allowed their service to lapse.

```python
# Logic for defining the proxy/label
def define_churn_label(days_since_last_active):
    return 1 if days_since_last_active > 30 else 0

print(f"Label for 45 days inactivity: {define_churn_label(45)}")
```

## 3. Success metric

**Metric:** F1-Score (specifically for the 'Churn' class).
**Success Threshold:** An **F1-score > 0.65**. Since churners are the minority class, Accuracy is misleading. We need to balance Precision (don't annoy loyal customers with retention offers) and Recall (capture as many potential churners as possible).

```python
from sklearn.metrics import f1_score
# Mock evaluation
y_true = [0, 1, 0, 0, 1, 0]
y_pred = [0, 1, 0, 0, 0, 0]
print(f"Current F1-Score: {f1_score(y_true, y_pred):.2f}")
```

## 4. The unit of analysis, as a real dataframe

**Unit of Analysis:** One row = **One Customer ID** at a specific point in time (e.g., the end of a billing cycle).

```python
# Loading a sample slice
data = {
    'customer_id': ['C001', 'C002', 'C003'],
    'monthly_usage_hrs': [45, 10, 120],
    'support_tickets': [0, 5, 1],
    'is_churned': [0, 1, 0]
}
df = pd.DataFrame(data)
display(df)
```

## 5. Why ML beats a fixed rule here

A simple rule like `if support_tickets > 3 then churn` is too rigid. Some high-usage users might have many tickets but remain loyal because they rely on the product. ML can weigh `usage_hrs` against `support_tickets` and `tenure` simultaneously, capturing the **non-linear relationship** between engagement and dissatisfaction that a single threshold cannot.

```python
# Demonstrating complexity
# A customer with many tickets (5) but high usage (120) might stay,
# while one with moderate tickets (2) and low usage (5) might leave.
print("Rule-based would fail on high-value, high-friction power users.")
```
